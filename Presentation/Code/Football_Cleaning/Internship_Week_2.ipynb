```python
import pandas as pd
```

```python
# Load the 2024-25 season data
df_2024_25 = pd.read_csv('/content/2024-25_E0.csv')

# Load the 2023-24 season data
df_2023_24 = pd.read_csv('/content/2023-24_E0.csv')

print('First 5 rows of 2024-25_E0.csv:')
display(df_2024_25.head())

print('\nFirst 5 rows of 2023-24_E0.csv:')
display(df_2023_24.head())
```

## Resources

- `FTHG`: Full Time Home Goals
- `HTHG`: Half Time Home Goals
- `FTAG`: Full Time Away Goals
- `HTAG`: Half Time Away Goals

```python
def clean_team_name(name):
    return name.replace(' ', '').lower()

# Apply cleaning to both DataFrames
df_2024_25['HomeTeam_Clean'] = df_2024_25['HomeTeam'].apply(clean_team_name)
df_2024_25['AwayTeam_Clean'] = df_2024_25['AwayTeam'].apply(clean_team_name)
df_2023_24['HomeTeam_Clean'] = df_2023_24['HomeTeam'].apply(clean_team_name)
df_2023_24['AwayTeam_Clean'] = df_2023_24['AwayTeam'].apply(clean_team_name)

# Get all unique team names from both seasons after initial cleaning
all_teams_2024_25 = pd.concat([df_2024_25['HomeTeam_Clean'], df_2024_25['AwayTeam_Clean']]).unique()
all_teams_2023_24 = pd.concat([df_2023_24['HomeTeam_Clean'], df_2023_24['AwayTeam_Clean']]).unique()

print("Unique team names after initial cleaning (2024-25 season):\n", sorted(all_teams_2024_25))
print("\nUnique team names after initial cleaning (2023-24 season):\n", sorted(all_teams_2023_24))
```

```python
# Define a comprehensive mapping for team names based on common variations and potential inconsistencies
team_name_mapping = {
    'arsenal': 'Arsenal',
    'astonvilla': 'Aston Villa',
    'bournemouth': 'Bournemouth',
    'brentford': 'Brentford',
    'brighton': 'Brighton',
    'burnley': 'Burnley',
    'chelsea': 'Chelsea',
    'crystalpalace': 'Crystal Palace',
    'everton': 'Everton',
    'fulham': 'Fulham',
    'ipswich': 'Ipswich Town',
    'leeds': 'Leeds United', # Example of a team that might appear in older datasets or could be misspelled
    'leicester': 'Leicester City',
    'liverpool': 'Liverpool',
    'luton': 'Luton Town',
    'manchesterunited': 'Man United', # Preferring 'Man United'
    'manunited': 'Man United',
    'manchestercity': 'Man City', # Preferring 'Man City'
    'mancity': 'Man City',
    'newcastle': 'Newcastle United',
    "nott'mforest": 'Nottingham Forest',
    'nottmforest': 'Nottingham Forest',
    'nottinghamforest': 'Nottingham Forest',
    'sheffieldunited': 'Sheffield United',
    'southampton': 'Southampton',
    'spurs': 'Tottenham',
    'tottenham': 'Tottenham',
    'westham': 'West Ham',
    'wolves': 'Wolves'
}

# Apply the mapping to both DataFrames
df_2024_25['HomeTeam_Standardized'] = df_2024_25['HomeTeam_Clean'].map(team_name_mapping).fillna(df_2024_25['HomeTeam_Clean'])
df_2024_25['AwayTeam_Standardized'] = df_2024_25['AwayTeam_Clean'].map(team_name_mapping).fillna(df_2024_25['AwayTeam_Clean'])
df_2023_24['HomeTeam_Standardized'] = df_2023_24['HomeTeam_Clean'].map(team_name_mapping).fillna(df_2023_24['HomeTeam_Clean'])
df_2023_24['AwayTeam_Standardized'] = df_2023_24['AwayTeam_Clean'].map(team_name_mapping).fillna(df_2023_24['AwayTeam_Clean'])

# Display unique standardized team names to confirm
all_standardized_teams_2024_25 = pd.concat([df_2024_25['HomeTeam_Standardized'], df_2024_25['AwayTeam_Standardized']]).unique()
all_standardized_teams_2023_24 = pd.concat([df_2023_24['HomeTeam_Standardized'], df_2023_24['AwayTeam_Standardized']]).unique()

print("Unique standardized team names (2024-25 season):\n", sorted(all_standardized_teams_2024_25))
print("\nUnique standardized team names (2023-24 season):\n", sorted(all_standardized_teams_2023_24))
```

```python
# Identify goal columns. Assuming 'FTHG', 'FTAG', 'HTHG', 'HTAG' are the main goal columns.
goal_columns = ['FTHG', 'FTAG', 'HTHG', 'HTAG']

print("Null values before flagging (2024-25 season):\n")
for col in goal_columns:
    df_2024_25[f'{col}_IsNull'] = df_2024_25[col].isnull()
    print(f"  {col}: {df_2024_25[col].isnull().sum()} nulls")

print("\nNull values before flagging (2023-24 season):\n")
for col in goal_columns:
    df_2023_24[f'{col}_IsNull'] = df_2023_24[col].isnull()
    print(f"  {col}: {df_2023_24[col].isnull().sum()} nulls")

print("\nExample of flagged data (2024-25 season, first 5 rows with goal columns and flags):\n")
display(df_2024_25[['FTHG', 'FTHG_IsNull', 'FTAG', 'FTAG_IsNull', 'HTHG', 'HTHG_IsNull', 'HTAG', 'HTAG_IsNull']].head())

print("\nExample of flagged data (2023-24 season, first 5 rows with goal columns and flags):\n")
display(df_2023_24[['FTHG', 'FTHG_IsNull', 'FTAG', 'FTAG_IsNull', 'HTHG', 'HTHG_IsNull', 'HTAG', 'HTAG_IsNull']].head())
```

```python
# Update original HomeTeam and AwayTeam columns with standardized names
df_2024_25['HomeTeam'] = df_2024_25['HomeTeam_Standardized']
df_2024_25['AwayTeam'] = df_2024_25['AwayTeam_Standardized']
df_2023_24['HomeTeam'] = df_2023_24['HomeTeam_Standardized']
df_2023_24['AwayTeam'] = df_2023_24['AwayTeam_Standardized']

# Drop the intermediate and standardized columns, keeping only the updated original names
df_2024_25.drop(columns=['HomeTeam_Standardized', 'AwayTeam_Standardized'], inplace=True)
df_2023_24.drop(columns=['HomeTeam_Standardized', 'AwayTeam_Standardized'], inplace=True)

print('Updated DataFrames with standardized team names:')
print('2024-25 season (first 5 rows):')
display(df_2024_25[['HomeTeam', 'AwayTeam']].head())
print('\n2023-24 season (first 5 rows):')
display(df_2023_24[['HomeTeam', 'AwayTeam']].head())
```

### Usual Cleaning Process: Checking Missing Values and Data Types

```python
# Check for missing values across all columns for both DataFrames
print("Missing values in 2024-25 season data:")
df_2024_25_missing = df_2024_25.isnull().sum()
display(df_2024_25_missing[df_2024_25_missing > 0])

print("\nMissing values in 2023-24 season data:")
df_2023_24_missing = df_2023_24.isnull().sum()
display(df_2023_24_missing[df_2023_24_missing > 0])

# Convert 'Date' column to datetime objects
df_2024_25['Date'] = pd.to_datetime(df_2024_25['Date'], format='%d/%m/%Y')
df_2023_24['Date'] = pd.to_datetime(df_2023_24['Date'], format='%d/%m/%Y')

# Combine 'Date' and 'Time' to create a single datetime column
df_2024_25['DateTime'] = df_2024_25.apply(lambda row: pd.to_datetime(f"{row['Date'].strftime('%Y-%m-%d')} {row['Time']}"), axis=1)
df_2023_24['DateTime'] = df_2023_24.apply(lambda row: pd.to_datetime(f"{row['Date'].strftime('%Y-%m-%d')} {row['Time']}"), axis=1)

print('\nData types after conversion for 2024-25 season:')
display(df_2024_25[['Date', 'Time', 'DateTime']].info())
print('\nData types after conversion for 2023-24 season:')
display(df_2023_24[['Date', 'Time', 'DateTime']].info())
```

### Combine DataFrames and Export to CSV

```python
# Identify common columns between the two dataframes to ensure a clean concatenation
common_columns = list(set(df_2024_25.columns) & set(df_2023_24.columns))

# Select only common columns for concatenation
df_2024_25_common = df_2024_25[common_columns]
df_2023_24_common = df_2023_24[common_columns]

# Concatenate the two DataFrames
combined_df = pd.concat([df_2024_25_common, df_2023_24_common], ignore_index=True)

print('Combined DataFrame info:')
display(combined_df.info())

print('\nFirst 5 rows of the combined DataFrame:')
display(combined_df.head())

# Export the combined DataFrame to a CSV file
output_filename = 'combined_epl_seasons_cleaned.csv'
combined_df.to_csv(output_filename, index=False)

print(f"\nCombined and cleaned data exported to {output_filename}")
```

### Sanity Check: Comparing Wins, Draws, Losses with Wikipedia Data

```python
# Load the Wikipedia league table data
df_wiki = pd.read_csv('/content/wikipedia_league_tables.csv')

print("First 5 rows of Wikipedia league table data:")
display(df_wiki.head())

# Perform initial cleaning on 'team' column in Wikipedia data for consistency
df_wiki['team_cleaned'] = df_wiki['team'].apply(lambda x: x.replace(' ', '').lower())
df_wiki['team_standardized'] = df_wiki['team_cleaned'].map(team_name_mapping).fillna(df_wiki['team_cleaned'])

# Standardize season format in df_wiki (replace en-dash with hyphen)
df_wiki['season'] = df_wiki['season'].str.replace('–', '-')

# Ensure the 'season' column is in a comparable format (e.g., extract year from 'season' string if needed)
# Assuming 'season' column in df_wiki is already in a consistent format like '2023-24'
# If not, further parsing would be needed, for now, we'll assume it's directly comparable or use integer years.
# For our combined_df, we'll derive the season from the 'Date' column.

print("\nUnique teams in Wikipedia data after standardization:")
print(sorted(df_wiki['team_standardized'].unique()))
```

```python
# Extract season from combined_df's 'Date' column
combined_df['Season'] = combined_df['Date'].dt.year.astype(str) + '-' + (combined_df['Date'].dt.year + 1).astype(str).str[2:]

# Initialize columns for aggregation
combined_df['HomeWin'] = (combined_df['FTR'] == 'H').astype(int)
combined_df['AwayWin'] = (combined_df['FTR'] == 'A').astype(int)
combined_df['Draw'] = (combined_df['FTR'] == 'D').astype(int)

# Aggregate wins, draws, losses for each team for each season from combined_df
team_stats = []

for season in combined_df['Season'].unique():
    season_df = combined_df[combined_df['Season'] == season]

    # Home team stats
    home_stats = season_df.groupby('HomeTeam').agg(
        won=('HomeWin', 'sum'),
        drawn=('Draw', 'sum'),
        lost=('AwayWin', 'sum') # If home team didn't win, and it wasn't a draw, it's a loss
    ).reset_index()
    home_stats.rename(columns={'HomeTeam': 'team', 'won': 'won_home', 'drawn': 'drawn_home', 'lost': 'lost_home'}, inplace=True)

    # Away team stats
    away_stats = season_df.groupby('AwayTeam').agg(
        won=('AwayWin', 'sum'),
        drawn=('Draw', 'sum'),
        lost=('HomeWin', 'sum') # If away team didn't win, and it wasn't a draw, it's a loss
    ).reset_index()
    away_stats.rename(columns={'AwayTeam': 'team', 'won': 'won_away', 'drawn': 'drawn_away', 'lost': 'lost_away'}, inplace=True)

    # Merge home and away stats
    merged_stats = pd.merge(home_stats, away_stats, on='team', how='outer').fillna(0)

    # Calculate total stats
    merged_stats['won'] = merged_stats['won_home'] + merged_stats['won_away']
    merged_stats['drawn'] = merged_stats['drawn_home'] + merged_stats['drawn_away']
    merged_stats['lost'] = merged_stats['lost_home'] + merged_stats['lost_away']
    merged_stats['played'] = merged_stats['won'] + merged_stats['drawn'] + merged_stats['lost']

    merged_stats['Season'] = season
    team_stats.append(merged_stats[['Season', 'team', 'played', 'won', 'drawn', 'lost']])

aggregated_combined_df = pd.concat(team_stats, ignore_index=True)

print("Aggregated stats from combined_df:")
display(aggregated_combined_df.head())

# Prepare df_wiki for merging
# Filter wiki data to only include seasons present in our combined_df
df_wiki_filtered = df_wiki[df_wiki['season'].isin(aggregated_combined_df['Season'].unique())]

# Merge and compare
comparison_df = pd.merge(
    aggregated_combined_df,
    df_wiki_filtered[['season', 'team_standardized', 'won', 'drawn', 'lost', 'played']],
    left_on=['Season', 'team'],
    right_on=['season', 'team_standardized'],
    suffixes=('_combined', '_wiki'),
    how='left'
)

# Check for discrepancies
discrepancies = comparison_df[
    (comparison_df['won_combined'] != comparison_df['won_wiki']) |
    (comparison_df['drawn_combined'] != comparison_df['drawn_wiki']) |
    (comparison_df['lost_combined'] != comparison_df['lost_wiki']) |
    (comparison_df['played_combined'] != comparison_df['played_wiki'])
]

if discrepancies.empty:
    print("\nSanity check passed: Wins, draws, and losses match between combined_df and Wikipedia data for the available seasons.")
else:
    print("\nSanity check failed: Discrepancies found between combined_df and Wikipedia data:")
    display(discrepancies)
```
